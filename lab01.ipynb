# 1. Virtual Programming Environment

In the name of reproducibility (and minimizing the number of headaches you get while doing your work), it's always a good idea to do science in a self-contained programming environment. This is essentially a directory in your computer's file system where isolated collections of programming libraries, in addition to computing language interpreters (in our case Python) live. This will minimize program library dependency issues within your machine. This can also come in handy if you ever write a bit of code you'd like someone else to be able to run on their machine. Yay interpretability!

## Command Line installation

Copy and paste these commands into your command line interface to initiate downloading miniconda3.

### **Windows**: 

Navigate to your command line interface. There are a few options here. I will show you how to install using the **Command Prompt**. Please refer [here](https://www.anaconda.com/docs/getting-started/miniconda/install/windows-cli-install) if you would rather use **PowerShell**.

```python
# Copy and paste the following command into your terminal. "curl" is a command line tool for downloading data with URLs.
curl https://repo.anaconda.com/miniconda/Miniconda3-latest-Windows-x86_64.exe --output .\Miniconda3-latest-Windows-x86_64.exe
```

Then, navigate to Home folder to double-click the installer to launch. Follow the prompts for installation. You should select the "Just Me" installation option. I would recommend storing Miniconda3 in your base user directory (e.g. C:\Users\USERNAME).

After this is finished, navigate to "Anaconda Prompt" from your search bar. This is where you can interface with miniconda from now on.

### **MacOS**: 

Hit command + spacebar and type Terminal into the spotlight bar. The terminal window will open up. After which you should copy and paste the following. Refer to [this link](https://www.anaconda.com/docs/getting-started/miniconda/install/mac-cli-install) for further info if needed: 

```python
# this downloads the miniconda installer
curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-MacOSX-arm64.sh
# this command runs the installer. FYI: the .sh extension indicates that this is a shell script.
bash ~/Miniconda3-latest-MacOSX-arm64.sh
```

Then quit and reopen your terminal. You should see "(base)" before your typical terminal header. If not, please write "conda activate" and hit Return.

### **Linux**:

Open up your terminal and paste the following:

```python
# This downloads the miniconda installer
curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
# This installs miniconda
bash ~/Miniconda3-latest-Linux-x86_64.sh
```

As above, close and reopen terminal to ensure conda is available.

### **For ALL:** Installing the libmamba dependency solver and setting it as primary solver.

You might find that installing packages in the future taking *forever*. To mitigate that, you'll install the libmamba dependency solver. The tl;dr on this solver is that it's written in a programming language that resolves package dependency issues faster than if you were to do this in a Python-centric workflow. Follow the instructions below. Make sure you see "(base)" at the beginning of your terminal! 

```python
conda install -n base conda-libmamba-solver
conda config --set solver libmamba
```
